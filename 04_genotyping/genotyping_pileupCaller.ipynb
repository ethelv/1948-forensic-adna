```python
# Jupyter Notebook cell

import os

# Define paths
bam_file = "JD206_filtered_filtered.bam"
reference_genome = "/Users/ethelvol/research/resources/hs37d5.fa"  # Replace with your fasta reference
snp_file = "v54.1.p1_HO_public.snp"                 # Replace with your Eigenstrat SNP file
sample_name = "JD206"
output_prefix = "JD206_filtered_filtered"

# Step 1: Generate pileup
pileup_cmd = (
    f"samtools mpileup -R -B -q30 -Q30 -f {reference_genome} {bam_file} > pileup.txt"
)

# Step 2: Run pileupCaller
pileup_caller_cmd = (
    f"pileupCaller --randomHaploid --sampleNames {sample_name} "
    f"-f {snp_file} -e {output_prefix} < pileup.txt"
)

print("Running samtools mpileup...")
os.system(pileup_cmd)

print("Running pileupCaller...")
os.system(pileup_caller_cmd)

print("Done! Output files are:", f"{output_prefix}.geno, {output_prefix}.ind, {output_prefix}.pos")
```
